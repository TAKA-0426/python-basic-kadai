```python
class Human:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name} は大人です")
        else:
            print(f"{self.name} は大人ではありません")


# インスタンスを複数作成してリストに格納
humans = [
    Human("田中", 25),
    Human("佐藤", 18),
    Human("鈴木", 20)
]

# リストの要素数分だけメソッドを呼び出す
for human in humans:
    human.check_adult()
```
